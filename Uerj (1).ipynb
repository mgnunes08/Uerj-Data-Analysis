```python
# MODIFICAÇÕES NO DATAFRAME

import pandas as pd
import matplotlib.pyplot as plt
dados1eq=pd.read_csv("uerj_finalcsv")
dados2eq=pd.read_csv("uerj_final2csv")
df1eq=pd.DataFrame(dados1eq)
df2eq=pd.DataFrame(dados2eq)
df1=pd.DataFrame(df1eq['Conceito'].value_counts().reset_index())
df1.columns=['Conceito','Quantidade']
df2eq.columns=['Inscrição','Nome','Acertos','Conceito']
df2=pd.DataFrame(df2eq['Conceito'].value_counts().reset_index())
df2.columns=['Conceito','Quantidade']
df1.drop([5,6],axis=0,inplace=True)

```

```python
# QUANTIDADE DE ALUNOS PRESENTES POR EXAME

qtd1eq=df1['Quantidade'].sum()
qtd2eq=df2['Quantidade'].sum()
eixox=['EQ1','EQ2']
eixoy=[qtd1eq,qtd2eq]
plt.bar(eixox,eixoy)
plt.title('Alunos presentes por exame',fontsize=20)
plt.show()
```

```python
# QUANTIDADE DE ALUNOS POR CONCEITO NOS DOIS EXAMES

fig=plt.figure()
axes=fig.add_axes([0.1,1,0.8,0.8])
axes2=fig.add_axes([1,1,0.8,0.8])
axes.bar(df1['Conceito'],df1['Quantidade'])
axes.set_title("EQ1")
axes2.bar(df2['Conceito'],df2['Quantidade'],color='darkorange')
plt.title("EQ2")
plt.show()
```

```python
import numpy as np
a1 = pd.to_numeric(df1eq['Acertos'], errors='coerce').dropna()
a2 = pd.to_numeric(df2eq['Acertos'], errors='coerce').dropna()
fig,ax=plt.subplots(nrows=1,ncols=2,figsize=(10,5))
ax[0].hist(a1,bins = np.arange(-0.5, 61.5, 1))
ax[0].set_title('EQ1')
ax[1].hist(a2,bins = np.arange(-0.5, 61.5, 1),color='darkorange')
ax[1].set_title('EQ2')
plt.show()
```
